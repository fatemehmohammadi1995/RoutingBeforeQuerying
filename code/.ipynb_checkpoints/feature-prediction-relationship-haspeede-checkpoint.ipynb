# Relationship between Text Features and Performance Prediction for Italian Hate-Speech Classification

**Future work:** In the future, we are going to use this relationship to train a router to select the best model for each example based on its features in workflows like an LLM chain. 

The central hypothesis is:

> Observable linguistic, semantic, stylistic, and structural characteristics
> of an input can help predict which LLM–prompt configuration is most
> appropriate for classifying that input.

The experimental design consists of:

- 200 Mappa dell'intolleranza human annotated Italian examples
- linguistic/pragmatic feature extraction
- EDA using extracted features
- Prediction using Qwen and Gemma models
- zero-shot, few-shot, and definition+few-shot prompting
- 6 LLM–prompt configurations
- one classification task:
  - 4-class label classification
- stratified 3-fold cross-validation
- Macro-F1 as the primary performance metric
- latency, token usage, invalid-output rate

This is intentionally a small proof-of-concept study. Larger datasets,
more models, multilingual experiments, more sophisticated routing methods,
and broader fairness analyses are reserved for the future journal study.

# 5. Feature extraction functions

```python
# ============================================================
# READABILITY
# ============================================================

def gulpease_index(text: str) -> float:
    """
    Gulpease readability index for Italian.

    Higher values generally indicate easier readability.
    """
    chars = len(re.sub(r"\s", "", text))
    words = len(text.split())

    if words == 0:
        return 0.0

    sentences = max(
        len(re.findall(r"[.!?]", text)),
        1,
    )

    return (
        89
        + (300 * sentences - 10 * chars) / words
    )


def count_syllables(word: str) -> int:
    vowels = "aeiouàèéìòù"

    return max(
        1,
        len(
            re.findall(
                f"[{vowels}]+",
                word.lower()
            )
        ),
    )


def flesch_vacca(text: str) -> float:
    """
    Flesch-Vacca readability index,
    an Italian adaptation of Flesch readability.
    """

    words = text.split()

    if not words:
        return 0.0

    syllables = sum(
        count_syllables(w)
        for w in words
    )

    sentences = max(
        len(re.findall(r"[.!?]", text)),
        1,
    )

    return (
        206
        - 0.65 * (len(words) / sentences)
        - 1.0 * (syllables / len(words))
    )
```

```python
# ============================================================
# FEATURE EXTRACTOR
# ============================================================

class FeatureExtractor:

    def __init__(
        self,
        spacy_model="it_core_news_sm",
        use_transformer_sentiment=False,
        sentiment_model_name="MilaNLProc/feel-it-italian-sentiment",
    ):

        import spacy

        self.nlp = spacy.load(spacy_model)

        self.use_transformer_sentiment = (
            use_transformer_sentiment
        )

        self.sentiment_pipeline = None

        if use_transformer_sentiment:

            try:

                from transformers import pipeline

                self.sentiment_pipeline = pipeline(
                    "sentiment-analysis",
                    model=sentiment_model_name,
                )

            except Exception as e:

                print(
                    "[FeatureExtractor] "
                    "Could not load transformer sentiment model."
                )

                print(
                    "Falling back to lexicon sentiment."
                )

                print("Reason:", e)

                self.use_transformer_sentiment = False

    # --------------------------------------------------------
    # SENTIMENT
    # --------------------------------------------------------

    def _sentiment(self, text):

        if (
            self.use_transformer_sentiment
            and self.sentiment_pipeline is not None
        ):

            try:

                output = self.sentiment_pipeline(
                    text[:512]
                )[0]

                return {
                    "sentiment_label": output["label"],
                    "sentiment_score": output["score"],
                }

            except Exception:

                return {
                    "sentiment_label": "unknown",
                    "sentiment_score": 0.0,
                }

        # ----------------------------------------------------
        # Offline lexicon fallback
        # ----------------------------------------------------

        tokens = re.findall(
            r"\w+",
            text.lower(),
        )

        positive_hits = sum(
            t in _POS_LEXICON
            for t in tokens
        )

        negative_hits = sum(
            t in _NEG_LEXICON
            for t in tokens
        )

        total = (
            positive_hits
            + negative_hits
        )

        if total == 0:

            return {
                "sentiment_label": "neutral",
                "sentiment_score": 0.0,
            }

        polarity = (
            positive_hits - negative_hits
        ) / total

        if polarity > 0:
            label = "positive"

        elif polarity < 0:
            label = "negative"

        else:
            label = "neutral"

        return {
            "sentiment_label": label,
            "sentiment_score": abs(polarity),
        }

    # --------------------------------------------------------
    # SINGLE TEXT
    # --------------------------------------------------------

    def extract(self, text):

        text = (
            ""
            if text is None
            else str(text)
        )

        doc = self.nlp(text)

        tokens = [
            t
            for t in doc
            if not t.is_punct
            and not t.is_space
        ]

        words = [
            t.text.lower()
            for t in tokens
        ]

        word_count = len(words)

        features = {}

        # ----------------------------------------------------
        # Basic text statistics
        # ----------------------------------------------------

        features["text_length_chars"] = len(text)

        features["text_length_words"] = word_count

        unique_words = len(set(words))

        features["lexical_diversity"] = (
            unique_words / word_count
            if word_count
            else 0.0
        )

        # ----------------------------------------------------
        # Readability
        # ----------------------------------------------------

        features["gulpease_index"] = (
            gulpease_index(text)
        )

        features["flesch_vacca_index"] = (
            flesch_vacca(text)
        )

        # ----------------------------------------------------
        # Linguistic sophistication
        # ----------------------------------------------------

        word_lengths = [
            len(w)
            for w in words
        ]

        features["avg_word_length"] = (
            float(np.mean(word_lengths))
            if words
            else 0.0
        )

        long_words = [
            w
            for w in words
            if len(w) >= 8
        ]

        features["long_word_ratio"] = (
            len(long_words) / word_count
            if word_count
            else 0.0
        )

        # ----------------------------------------------------
        # Syntactic complexity
        # ----------------------------------------------------

        sentence_lengths = [
            len(sent)
            for sent in doc.sents
        ]

        features["avg_sentence_length"] = (
            float(np.mean(sentence_lengths))
            if sentence_lengths
            else 0.0
        )

        dependency_depth = []

        for token in doc:

            depth = 0
            child = token

            while child.head != child:

                depth += 1

                child = child.head

                if depth > 200:
                    break

            dependency_depth.append(depth)

        features["dependency_tree_depth"] = (
            float(np.mean(dependency_depth))
            if dependency_depth
            else 0.0
        )

        # ----------------------------------------------------
        # Social media markers
        # ----------------------------------------------------

        features["hashtag_count"] = len(
            re.findall(
                r"#\w+",
                text
            )
        )

        features["mention_count"] = len(
            re.findall(
                r"@\w+",
                text
            )
        )

#         features["url_count"] = len(
#             re.findall(
#                 r"https?://\S+",
#                 text
#             )
#         )

        try:

            import emoji

            features["emoji_count"] = len(
                [
                    c
                    for c in text
                    if c in emoji.EMOJI_DATA
                ]
            )

        except Exception:

            features["emoji_count"] = len(
                re.findall(
                    r"[\U0001F300-\U0001FAFF]",
                    text
                )
            )

        # ----------------------------------------------------
        # Sentiment
        # ----------------------------------------------------

        features.update(
            self._sentiment(text)
        )

        # ----------------------------------------------------
        # Offensive language
        # ----------------------------------------------------

        features["offensive_word_count"] = sum(
            w in OFFENSIVE_WORDS
            for w in words
        )

        # ----------------------------------------------------
        # Pragmatics
        # ----------------------------------------------------

        features["second_person_ratio"] = (
            sum(
                w in SECOND_PERSON
                for w in words
            ) / word_count
            if word_count
            else 0.0
        )

        features["negation_count"] = sum(
            w in NEGATIONS
            for w in words
        )

        # ----------------------------------------------------
        # Emotional / emphatic writing
        # ----------------------------------------------------

        raw_tokens = text.split()

        upper_words = [
            w
            for w in raw_tokens
            if w.isupper()
            and len(w) > 1
        ]

        features["uppercase_ratio"] = (
            len(upper_words) / len(raw_tokens)
            if raw_tokens
            else 0.0
        )

        elongated = re.findall(
            r"(.)\1{2,}",
            text.lower()
        )

        features["elongated_word_count"] = (
            len(elongated)
        )

        # ----------------------------------------------------
        # POS distribution
        # ----------------------------------------------------

        pos_counts = Counter(
            token.pos_
            for token in tokens
        )

        for pos, count in pos_counts.items():

            features[
                f"POS_{pos}_ratio"
            ] = (
                count / word_count
                if word_count
                else 0.0
            )

        # ----------------------------------------------------
        # Named entities
        # ----------------------------------------------------

        features["named_entity_count"] = len(
            doc.ents
        )

        features["unique_entity_types"] = len(
            set(
                ent.label_
                for ent in doc.ents
            )
        )

        # ----------------------------------------------------
        # Sarcasm proxies
        # ----------------------------------------------------

        lower = text.lower()

        features["sarcasm_marker_count"] = sum(
            marker in lower
            for marker in SARCASM_MARKERS
        )

        features["sarcasm_emoji_count"] = sum(
            text.count(e)
            for e in SARCASM_EMOJIS
        )

        features["exclamation_count"] = (
            text.count("!")
        )

        features[
            "question_exclamation_count"
        ] = len(
            re.findall(
                r"\?!+|!\?+",
                text
            )
        )

        features["quoted_words_count"] = len(
            re.findall(
                r'["«»](.*?)["«»]',
                text
            )
        )

        return features

    # --------------------------------------------------------
    # BATCH
    # --------------------------------------------------------

    def extract_batch(self, texts):

        rows = [
            self.extract(text)
            for text in texts
        ]

        result = pd.DataFrame(rows)

        pos_columns = [
            c
            for c in result.columns
            if c.startswith("POS_")
        ]

        if pos_columns:

            result[pos_columns] = (
                result[pos_columns]
                .fillna(0.0)
            )

        return result
```

```python
numeric_cols = feature_df.select_dtypes(
    include=np.number
).columns.tolist()

binary_cols = [
    col for col in numeric_cols
    if feature_df[col].nunique(dropna=False) <= 2
]

continuous_cols = [
    col for col in numeric_cols
    if col not in binary_cols
]

categorical_like_cols = [
    col for col in numeric_cols
    if feature_df[col].nunique(dropna=False) < 10
]

print(f"Total numerical features : {len(numeric_cols)}")
print(f"Binary features          : {len(binary_cols)}")
print(f"Continuous features      : {len(continuous_cols)}")
print(f"Low-cardinality features : {len(categorical_like_cols)}")
```

```python
# ============================================================
# 1. Descriptive statistics
# ============================================================

print("=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

stats = feature_df[numeric_cols].describe().T

stats["missing"] = feature_df[numeric_cols].isna().sum()
stats["n_unique"] = feature_df[numeric_cols].nunique()
stats["missing_pct"] = stats["missing"] / len(feature_df) * 100

display(
    stats.sort_values("std", ascending=False)
)
```

```python
# ============================================================
# 2. Constant and near-constant features
# ============================================================

print("=" * 70)
print("CONSTANT / NEAR-CONSTANT FEATURES")
print("=" * 70)

n_unique = feature_df.nunique()

constant_features = n_unique[n_unique <= 1].index.tolist()

print(f"Constant features: {len(constant_features)}")

if constant_features:
    display(
        feature_df[constant_features].head()
    )

# Proportion of the most common value
dominance = pd.DataFrame(index=numeric_cols)

for col in numeric_cols:
    dominance.loc[col, "n_unique"] = feature_df[col].nunique()
    dominance.loc[col, "most_common_pct"] = (
        feature_df[col].value_counts(normalize=True).iloc[0] * 100
    )

near_constant = dominance[
    dominance["most_common_pct"] >= 99
].sort_values(
    "most_common_pct",
    ascending=False
)

print(f"\nNear-constant features (>=99% same value): {len(near_constant)}")
display(near_constant)
```

```python
# ============================================================
# 3. Distribution of numerical features
# ============================================================

# Plot histograms in batches so large feature sets remain manageable.

MAX_FEATURES_PER_FIGURE = 20

for start in range(0, len(continuous_cols), MAX_FEATURES_PER_FIGURE):

    cols = continuous_cols[
        start:start + MAX_FEATURES_PER_FIGURE
    ]

    n = len(cols)
    ncols = 4
    nrows = int(np.ceil(n / ncols))

    fig, axes = plt.subplots(
        nrows,
        ncols,
        figsize=(16, 3.5 * nrows)
    )

    axes = np.atleast_1d(axes).flatten()

    for ax, col in zip(axes, cols):
        ax.hist(
            feature_df[col],
            bins=40,
            alpha=0.8
        )
        ax.set_title(col)
        ax.set_xlabel("Value")
        ax.set_ylabel("Count")

    # Hide unused axes
    for ax in axes[len(cols):]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```

# 7. Prompt definitions

We use three strategies:

1. Zero-shot
2. Few-shot
3. Definition + few-shot

Few-shot examples are selected exclusively from the corresponding
training fold.

No test examples are used as demonstrations.

```python
print("TEXT_COL:", TEXT_COL)
print("LABEL_COL:", LABEL_COL)

print(
    "LABEL_COL exists:",
    LABEL_COL in df.columns,
)

print(
    "Available columns:",
    df.columns.tolist(),
)
```

```python
# ============================================================
# LABEL DEFINITIONS
# ============================================================

LABEL_DEFINITIONS = {
    "neutral":
        "Content that does not express hate speech "
        "or a stereotype according to the annotation scheme.",

    "hate_speech":
        "Content expressing hostility, derogation, "
        "dehumanization, or hateful content toward "
        "a target group or person according to the annotation scheme.",

}

LABELS = list(LABEL_DEFINITIONS.keys())


# ============================================================
# FEW-SHOT EXAMPLE SELECTION
# ============================================================

def select_few_shot_examples(
    dataframe,
    target_column,
    n_per_class=2,
    seed=RANDOM_SEED,
):
    """
    Select a fixed, balanced set of few-shot examples.

    Returns:
        list[dict]
    """

    examples = []

    for class_name, group in dataframe.groupby(
        target_column
    ):

        n = min(
            n_per_class,
            len(group),
        )

        sampled = group.sample(
            n=n,
            random_state=seed,
        )

        examples.extend(
            sampled.to_dict("records")
        )

    return examples
```

```python
# ============================================================
# LABEL PROMPT
# ============================================================

def build_label_prompt(
    text,
    strategy,
    few_shot_examples=None,
):

    intro = """
You are an Italian-language content classification system.

Classify the input according to the following annotation scheme.
Return exactly one label and nothing else.
""".strip()

    definitions = "\n".join(
        f"- {label}: {description}"
        for label, description
        in LABEL_DEFINITIONS.items()
    )

    allowed_labels = "\n".join(
        f"- {label}"
        for label in LABELS
    )

    # ========================================================
    # ZERO-SHOT
    # ========================================================

    if strategy == "zero_shot":

        return f"""
{intro}

Labels:
{definitions}

Input:
{text}

Return exactly one of:
{allowed_labels}
""".strip()

    # ========================================================
    # FEW-SHOT
    # ========================================================

    elif strategy == "few_shot":

        if few_shot_examples is None:
            raise ValueError(
                "few_shot_examples must be provided "
                "for few-shot prompting."
            )

        example_text = "\n\n".join(
            [
                (
                    f"Example:\n"
                    f"{example[TEXT_COL]}\n"
                    f"Label: {example[LABEL_COL]}"
                )
                for example in few_shot_examples
            ]
        )

        return f"""
{intro}

Allowed labels:
{allowed_labels}

Examples:

{example_text}

Input:
{text}

Return exactly one label.
""".strip()

    # ========================================================
    # DEFINITION + FEW-SHOT
    # ========================================================

    elif strategy == "definition_few_shot":

        if few_shot_examples is None:
            raise ValueError(
                "few_shot_examples must be provided "
                "for definition_few_shot prompting."
            )

        example_text = "\n\n".join(
            [
                (
                    f"Example:\n"
                    f"{example[TEXT_COL]}\n"
                    f"Label: {example[LABEL_COL]}"
                )
                for example in few_shot_examples
            ]
        )

        return f"""
{intro}

Operational definitions:
{definitions}

Examples:

{example_text}

Input:
{text}

Return exactly one label.
""".strip()

    else:

        raise ValueError(
            f"Unknown prompt strategy: {strategy}"
        )
```

# 9. Local Qwen/Gemma inference

This implementation uses Hugging Face Transformers.

The generation is deterministic (`do_sample=False`) so that repeated
experiments are as reproducible as possible.

```python
# ============================================================
# HUGGING FACE GENERATOR
# ============================================================

import time
import torch


class HFGenerator:

    def __init__(
        self,
        model_id,
        hf_token=None,
    ):

        from transformers import (
            AutoTokenizer,
            AutoModelForCausalLM,
        )

        self.model_id = model_id
        self.hf_token = hf_token

        print(
            f"Loading tokenizer: {model_id}"
        )

        self.tokenizer = (
            AutoTokenizer.from_pretrained(
                model_id,
                token=hf_token,
            )
        )

        # ----------------------------------------------------
        # DTYPE
        # ----------------------------------------------------

        if torch.cuda.is_available():

            if torch.cuda.is_bf16_supported():
                dtype = torch.bfloat16
            else:
                dtype = torch.float16

        else:

            dtype = torch.float32

        # ----------------------------------------------------
        # MODEL
        # ----------------------------------------------------

        print(
            f"Loading model: {model_id}"
        )

        model_kwargs = {
            "token": hf_token,
            "torch_dtype": dtype,
        }

        if torch.cuda.is_available():
            model_kwargs["device_map"] = "auto"

        self.model = (
            AutoModelForCausalLM.from_pretrained(
                model_id,
                **model_kwargs,
            )
        )

        if not torch.cuda.is_available():
            self.model.to("cpu")

        self.model.eval()

        print(
            f"Successfully loaded: {model_id}"
        )

    def generate(
        self,
        prompt,
        max_new_tokens=20,
    ):

        messages = [
            {
                "role": "user",
                "content": prompt,
            }
        ]

        # ----------------------------------------------------
        # TOKENIZE
        # ----------------------------------------------------

        if getattr(
            self.tokenizer,
            "chat_template",
            None,
        ):

            inputs = (
                self.tokenizer.apply_chat_template(
                    messages,
                    add_generation_prompt=True,
                    return_tensors="pt",
                    return_dict=True,
                )
            )

        else:

            inputs = self.tokenizer(
                prompt,
                return_tensors="pt",
            )

        inputs = inputs.to(
            self.model.device
        )

        prompt_len = (
            inputs["input_ids"]
            .shape[-1]
        )

        # ----------------------------------------------------
        # GENERATION
        # ----------------------------------------------------

        start_time = time.perf_counter()

        with torch.no_grad():

            outputs = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
            )

        latency = (
            time.perf_counter()
            - start_time
        )

        # ----------------------------------------------------
        # DECODE ONLY NEW TOKENS
        # ----------------------------------------------------

        generated_tokens = (
            outputs[0][prompt_len:]
        )

        generated_text = (
            self.tokenizer.decode(
                generated_tokens,
                skip_special_tokens=True,
            )
        )

        return {
            "text": generated_text,
            "latency_sec": latency,
            "input_tokens": int(
                prompt_len
            ),
            "output_tokens": int(
                generated_tokens.shape[-1]
            ),
        }
```

# 10. Output normalization

LLMs sometimes return:

> `The answer is: hate_speech`

rather than exactly:

> `hate_speech`

The normalization procedure therefore accepts the exact label and a
conservative substring match.

Anything else is counted as `INVALID`.

Invalid outputs are **not silently removed** from evaluation.

```python
# ============================================================
# OUTPUT NORMALIZATION
# ============================================================

def normalize_prediction(
    raw_output,
    allowed_labels=LABELS,
):

    text = str(
        raw_output
    ).strip().lower()

    # Exact match first
    for label in allowed_labels:

        if text == label.lower():
            return label

    # Then search response
    for label in sorted(
        allowed_labels,
        key=len,
        reverse=True,
    ):

        if label.lower() in text:
            return label

    return "INVALID"
```

# 13. Run one configuration on one fold

This function is the fundamental experimental unit.

It records the raw prediction as well as normalized prediction,
latency, token usage, and errors.

```python
# ============================================================
# RUN ONE CONFIGURATION
# ============================================================

def run_configuration(
    generator,
    config_name,
    task,
    dataframe,
    few_shot_examples=None,
):

    model_name, prompt_strategy = (
        config_name.split("__", 1)
    )

    if task != "label":
        raise ValueError(
            f"Unsupported task: {task}"
        )

    gold_column = LABEL_COL

    rows = []

    for index, row in tqdm(
        dataframe.iterrows(),
        total=len(dataframe),
        desc=config_name,
        unit="sample",
    ):

        text = row[TEXT_COL]

        # ----------------------------------------------------
        # BUILD PROMPT
        # ----------------------------------------------------

        prompt = build_label_prompt(
            text=text,
            strategy=prompt_strategy,
            few_shot_examples=few_shot_examples,
        )

        # ----------------------------------------------------
        # GENERATE
        # ----------------------------------------------------

        try:

            result = generator.generate(
                prompt
            )

            raw_output = result["text"]

            prediction = normalize_prediction(
                raw_output,
                LABELS,
            )

            error = ""

        except Exception as e:

            raw_output = ""
            prediction = "INVALID"

            result = {
                "latency_sec": np.nan,
                "input_tokens": np.nan,
                "output_tokens": np.nan,
            }

            error = repr(e)

        # ----------------------------------------------------
        # STORE
        # ----------------------------------------------------

        rows.append(
            {
                "sample_id": int(
                    row["sample_id"]
                ),

                "dataset_index": int(
                    index
                ),

                "task": task,

                "config": config_name,

                "model": model_name,

                "prompt_strategy": prompt_strategy,

                "gold": row[gold_column],

                "prediction": prediction,

                "raw_output": raw_output,

                "latency_sec": result[
                    "latency_sec"
                ],

                "input_tokens": result[
                    "input_tokens"
                ],

                "output_tokens": result[
                    "output_tokens"
                ],

                "error": error,
            }
        )

    return pd.DataFrame(rows)
```

# 14. Three-fold configuration evaluation

This produces the raw predictions required for all subsequent analyses.

The same folds are used for all configurations.

```python
# ============================================================
# RUN ALL CONFIGURATIONS
# ============================================================

def run_experiment(
    dataframe,
    task,
    generators,
    few_shot_examples=None,
):

    all_results = []

    for config_name in CONFIGURATIONS:

        model_name = (
            config_name.split(
                "__",
                1,
            )[0]
        )

        if model_name not in generators:

            print(
                f"Skipping {config_name}: "
                f"generator '{model_name}' "
                f"not found."
            )

            continue

        print(
            "\n" + "=" * 70
        )

        print(
            f"Running configuration: "
            f"{config_name}"
        )

        print(
            "=" * 70
        )

        result = run_configuration(
            generator=generators[
                model_name
            ],

            config_name=config_name,

            task=task,

            dataframe=dataframe,

            few_shot_examples=(
                few_shot_examples
            ),
        )

        all_results.append(
            result
        )

    if not all_results:

        raise RuntimeError(
            "No configurations were executed."
        )

    return pd.concat(
        all_results,
        ignore_index=True,
    )
```

# 17. Configuration-level evaluation

Macro-F1 is the primary metric.

Accuracy and weighted-F1 are secondary.

Invalid outputs are retained and reported.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
)


def summarize_config_results(results):

    summaries = []

    grouped = results.groupby(
        [
            "task",
            "config",
            "model",
            "prompt_strategy",
        ]
    )

    for group_name, group in grouped:

        y_true = group["gold"].astype(str)
        y_pred = group["prediction"].astype(str)

        # Only evaluate valid predictions
        valid_mask = (
            group["prediction"].notna()
            & group["prediction"].astype(str).str.strip().ne("")
            & group["prediction"].ne("INVALID")
        )

        valid = group[valid_mask]

        y_true_valid = valid["gold"].astype(str)
        y_pred_valid = valid["prediction"].astype(str)

        if len(valid) > 0:

            accuracy = accuracy_score(
                y_true_valid,
                y_pred_valid,
            )

            precision, recall, f1, _ = (
                precision_recall_fscore_support(
                    y_true_valid,
                    y_pred_valid,
                    average="macro",
                    zero_division=0,
                )
            )

        else:

            accuracy = np.nan
            precision = np.nan
            recall = np.nan
            f1 = np.nan

        # Numeric metrics
        latency = pd.to_numeric(
            group["latency_sec"],
            errors="coerce",
        )

        input_tokens = pd.to_numeric(
            group["input_tokens"],
            errors="coerce",
        )

        output_tokens = pd.to_numeric(
            group["output_tokens"],
            errors="coerce",
        )

        summaries.append(
            {
                "task": group["task"].iloc[0],

                "config": group["config"].iloc[0],

                "model": group["model"].iloc[0],

                "prompt_strategy": group[
                    "prompt_strategy"
                ].iloc[0],

                "n_samples": len(group),

                "n_valid": len(valid),

                "n_invalid": (
                    len(group) - len(valid)
                ),

                "accuracy": accuracy,

                "macro_precision": precision,

                "macro_recall": recall,

                "macro_f1": f1,

                "mean_latency_sec": (
                    latency.mean()
                ),

                "mean_input_tokens": (
                    input_tokens.mean()
                ),

                "mean_output_tokens": (
                    output_tokens.mean()
                ),
            }
        )

    return pd.DataFrame(summaries)
```

# 18. Classification metric helper

```python
# ============================================================
# METRIC HELPER
# ============================================================

def classification_metrics(
    results,
):

    return {
        "accuracy": accuracy_score(
            results["gold"],
            results["prediction"],
        ),

        "macro_f1": f1_score(
            results["gold"],
            results["prediction"],
            average="macro",
            zero_division=0,
        ),

        "weighted_f1": f1_score(
            results["gold"],
            results["prediction"],
            average="weighted",
            zero_division=0,
        ),

        "invalid_rate": (
            results["prediction"]
            == "INVALID"
        ).mean(),

        "mean_latency_sec": (
            results["latency_sec"]
            .mean()
        ),
    }
```

# 19. Per-class performance

```python
# ============================================================
# PER-CLASS F1
# ============================================================

def per_class_f1(
    results,
):

    labels = sorted(
        results["gold"]
        .dropna()
        .unique()
    )

    scores = f1_score(
        results["gold"],
        results["prediction"],
        labels=labels,
        average=None,
        zero_division=0,
    )

    return pd.DataFrame(
        {
            "class": labels,
            "f1": scores,
        }
    )


def class_performance_disparity(
    results,
):

    table = per_class_f1(
        results
    )

    return (
        table["f1"].max()
        - table["f1"].min()
    )
```

# 20. Feature-Prediction Relationship Analysis

```python
def make_feature_matrix(feature_dataframe):
    X = (
        feature_dataframe
        .select_dtypes(include=[np.number])
        .copy()
    )
    X = (
        X
        .replace([np.inf, -np.inf], np.nan)
        .fillna(0.0)
    )
    return X


X_features = make_feature_matrix(feature_df)
X_features.index = df["sample_id"]

print("Feature matrix:", X_features.shape)
X_features.head()
```

```python
# ============================================================
# FEATURE DISCRETIZATION
# ============================================================

def discretize_feature(series, max_categorical_unique=10, min_bin_frac=0.05):
    """
    Turn a numeric feature into a categorical Series for relationship
    testing.

    - Low-cardinality features (<= 10 distinct values -- this covers
      count-style features such as url_count, mention_count,
      negation_count, named_entity_count, ...) are kept as their native
      integer categories, with any sparse tail (individual values
      making up less than `min_bin_frac` of the corpus) capped into a
      single '<value>+' bucket. This is what correctly turns
      url_count in {0, 1, 2, 3} into the categories {'0', '1', '2+'},
      rather than forcing a generic tertile split that would blur the
      0-vs-at-least-one-URL distinction that matters here.
    - Higher-cardinality (continuous-like) features are split into
      tertiles ('low' / 'mid' / 'high') by quantile, falling back to a
      median split if tertiles cannot be formed (e.g. too many tied
      values).
    """

    s = series.astype(float)
    nunique = s.nunique(dropna=True)

    if nunique <= max_categorical_unique:
        vc = s.value_counts(normalize=True).sort_index()
        keep_values = vc[vc >= min_bin_frac].index
        if len(keep_values) == 0:
            keep_values = [vc.index.max()]
        cap = max(keep_values)
        capped = np.where(
            s >= cap, f"{int(cap)}+", s.astype(int).astype(str)
        )
        return pd.Series(capped, index=s.index)

    try:
        bins = pd.qcut(s, q=3, labels=["low", "mid", "high"], duplicates="drop")
        counts = bins.value_counts(normalize=True, dropna=True)
        if len(counts) >= 2 and counts.min() >= min_bin_frac:
            return bins.astype(str)
    except Exception:
        pass

    # fallback: median split
    median = s.median()
    return pd.Series(np.where(s > median, "high", "low"), index=s.index)


feat_categorical = pd.DataFrame(
    {feat: discretize_feature(X_features[feat]) for feat in X_features.columns},
    index=X_features.index,
)

print("Discretized feature table:", feat_categorical.shape)
feat_categorical.head()
```

```python
# ============================================================
# BUILD THE MERGED PREDICTION + FEATURE TABLE
# ============================================================

feature_cols = [c for c in X_features.columns]

feat_lookup = X_features.copy()
feat_lookup.index.name = "sample_id"

merged = label_results.copy()
merged["correct"] = (
    merged["gold"].astype(str) == merged["prediction"].astype(str)
).astype(int)

print("merged predictions:", merged.shape)
print("feature columns:", len(feature_cols))
```

```python
# ============================================================
# PAIRWISE DISAGREEMENT SIGNIFICANCE TESTING
# ============================================================

from itertools import combinations
from scipy.stats import chi2_contingency

MIN_GROUP_N = 15  # minimum instances per "winner" group to run a test


def cramers_v(chi2_stat, n, table_shape):
    r, c = table_shape
    return float(np.sqrt(chi2_stat / (n * (min(r, c) - 1)))) if min(r, c) > 1 else np.nan


def test_feature_relationships(merged, feat_categorical, configurations, feature_cols):
    """
    For every configuration pair and every feature, test (on the
    disagreement instances) whether the feature's category is
    associated with which configuration is correct.

    Returns one row per (config_A, config_B, feature) that had enough
    data to test, sorted by p-value.
    """

    pivot_correct = merged.pivot_table(
        index="sample_id",
        columns="config",
        values="correct",
        aggfunc="first",
    )

    rows = []

    for a, b in combinations(configurations, 2):
        if a not in pivot_correct.columns or b not in pivot_correct.columns:
            continue

        sub = pivot_correct[[a, b]].dropna()
        disagree = sub[a] != sub[b]
        d = sub[disagree]

        a_wins = d[d[a] == 1].index
        b_wins = d[d[b] == 1].index

        if len(a_wins) < MIN_GROUP_N or len(b_wins) < MIN_GROUP_N:
            continue

        for feat in feature_cols:
            cat = feat_categorical[feat]

            cat_a = cat.reindex(a_wins).dropna()
            cat_b = cat.reindex(b_wins).dropna()

            if len(cat_a) < MIN_GROUP_N or len(cat_b) < MIN_GROUP_N:
                continue

            winner = pd.Series(
                ["A"] * len(cat_a) + ["B"] * len(cat_b),
                index=list(cat_a.index) + list(cat_b.index),
            )
            category = pd.concat([cat_a, cat_b])

            ct = pd.crosstab(category, winner)

            if ct.shape[0] < 2 or ct.shape[1] < 2:
                continue

            chi2_stat, p_value, dof, expected = chi2_contingency(ct)

            if (expected < 5).mean() > 0.2:
                # Cochran's rule of thumb: too many low-expected-count
                # cells for the chi-square approximation to be reliable.
                continue

            rows.append({
                "config_A": a,
                "config_B": b,
                "feature": feat,
                "n_A_wins": len(cat_a),
                "n_B_wins": len(cat_b),
                "n_disagree": len(cat_a) + len(cat_b),
                "chi2": chi2_stat,
                "dof": dof,
                "cramers_v": cramers_v(chi2_stat, len(cat_a) + len(cat_b), ct.shape),
                "p_value": p_value,
            })

    return pd.DataFrame(rows).sort_values("p_value").reset_index(drop=True)


relationship_results = test_feature_relationships(
    merged, feat_categorical, CONFIGURATIONS, feature_cols
)

print("Total (config pair, feature) tests run:", len(relationship_results))
relationship_results.head(20)
```

```python
# ============================================================
# FIGURE 1: CONFIGURATION PERFORMANCE
# ============================================================

def plot_configuration_performance(
    summary,
    task,
):

    plot_data = (
        summary
        .sort_values(
            "macro_f1",
            ascending=True,
        )
        .copy()
    )

    fig, ax = plt.subplots(
        figsize=(10, 6)
    )

    bars = ax.barh(
        plot_data["config"],
        plot_data["macro_f1"],
    )

    # --------------------------------------------------------
    # ADD VALUES TO BARS
    # --------------------------------------------------------

    ax.bar_label(
        bars,
        labels=[
            f"{value:.2f}"
            for value in plot_data["macro_f1"]
        ],
        padding=4,
        fontsize=10,
    )

    ax.set_xlabel(
        "Macro-F1"
    )

    ax.set_ylabel(
        "LLM–prompt configuration"
    )

    ax.set_title(
        f"{task.capitalize()} classification: "
        "configuration performance"
    )

    ax.set_xlim(
        0,
        min(
            1.0,
            plot_data["macro_f1"].max()
            + 0.10,
        ),
    )

    fig.tight_layout()

    path = (
        FIGURE_DIR
        / f"{task}_configuration_performance.pdf"
    )

    fig.savefig(
        path,
        bbox_inches="tight",
    )

    plt.show()

    return path
```

# 36. Feature distributions by class

These plots are exploratory.

They should not be interpreted as proving that a feature causes a particular
classification outcome.

Their purpose is to visually motivate the existence of heterogeneous
instance characteristics.

```python
# ============================================================
# FEATURE DISTRIBUTION PLOTS
# ============================================================

def plot_feature_by_class(
    dataframe,
    class_column,
    features,
    filename_prefix,
):

    generated_files = []

    for feature in features:

        if feature not in dataframe.columns:
            continue

        if not pd.api.types.is_numeric_dtype(
            dataframe[feature]
        ):
            continue

        plt.figure(
            figsize=(8, 5)
        )

        sns.boxplot(
            data=dataframe,
            x=class_column,
            y=feature,
        )

        plt.xticks(
            rotation=30,
            ha="right",
        )

        plt.xlabel(
            class_column
        )

        plt.ylabel(
            feature
        )

        plt.title(
            f"{feature} by {class_column}"
        )

        plt.tight_layout()

        path = (
            FIGURE_DIR
            / f"{filename_prefix}_{feature}.pdf"
        )

        plt.savefig(
            path,
            bbox_inches="tight",
        )

        plt.show()

        generated_files.append(
            path
        )

    return generated_files
```

# 44. Configuration-specific per-class results

```python
# ============================================================
# PER-CLASS TABLE FOR ALL CONFIGURATIONS
# ============================================================

def configuration_class_table(
    results,
):

    rows = []

    for config, group in (
        results.groupby("config")
    ):

        scores = per_class_f1(
            group
        )

        for _, row in scores.iterrows():

            rows.append(
                {
                    "config": config,
                    "class": row["class"],
                    "f1": row["f1"],
                }
            )

    return pd.DataFrame(
        rows
    )
```

```python
# import numpy as np
# import pandas as pd
# import matplotlib.pyplot as plt
# import seaborn as sns

# from sklearn.metrics import accuracy_score, f1_score


results = label_results.copy()
feat = X_features.copy()


# Move index into a normal column
feat = feat.reset_index()

print(feat.columns.tolist())

# Make sure IDs have the same type
results["sample_id"] = results["sample_id"].astype(str)
feat["sample_id"] = feat["sample_id"].astype(str)

# Remove duplicate feature rows if any
feat = feat.drop_duplicates("sample_id")

# Merge
df = results.merge(
    feat,
    on="sample_id",
    how="left",
    validate="many_to_one"
)

print("Inference rows:", len(results))
print("Merged rows:", len(df))
print("Unique samples:", df["sample_id"].nunique())

# ---------------------------------------------------------
# Correctness
# ---------------------------------------------------------

df["correct"] = (
    df["gold"].astype(str).str.strip().str.lower()
    ==
    df["prediction"].astype(str).str.strip().str.lower()
).astype(int)

# Remove failed / missing predictions from performance analyses
df_valid = df[
    df["prediction"].notna() &
    df["gold"].notna()
].copy()

print("\nConfigs:")
print(df_valid["config"].value_counts())

print("\nTasks:")
print(df_valid["task"].value_counts())

print("\nOverall performance:")
overall = (
    df_valid
    .groupby(["task", "config"])
    .apply(
        lambda x: pd.Series({
            "n": len(x),
            "accuracy": accuracy_score(x["gold"], x["prediction"]),
            "macro_f1": f1_score(
                x["gold"],
                x["prediction"],
                average="macro",
                zero_division=0
            )
        }),
        include_groups=False
    )
    .reset_index()
)

display(overall)
```

```python
# Features available for analysis
feature_cols = [
    c for c in feat.columns
    if c != "sample_id"
    and pd.api.types.is_numeric_dtype(feat[c])
]

print(f"{len(feature_cols)} numerical features")
print(feature_cols)
```

```python
N_BINS = 4

binned_data = df_valid.copy()

bin_info = {}

for feature in feature_cols:

    # Skip features with too few unique values
    n_unique = binned_data[feature].nunique(dropna=True)

    if n_unique < 4:
        print(f"Skipping {feature}: only {n_unique} unique values")
        continue

    try:
        binned_data[f"{feature}__bin"] = pd.qcut(
            binned_data[feature],
            q=N_BINS,
            duplicates="drop"
        )

        bin_info[feature] = f"{feature}__bin"

    except Exception as e:
        print(f"Could not bin {feature}: {e}")

print("Successfully binned:", len(bin_info))
```

```python
records = []

for feature, bin_col in bin_info.items():

    for task in binned_data["task"].dropna().unique():

        task_df = binned_data[
            binned_data["task"] == task
        ]

        for bin_value, bin_df in task_df.groupby(
            bin_col,
            observed=True
        ):

            for config, config_df in bin_df.groupby("config"):

                if len(config_df) == 0:
                    continue

                accuracy = accuracy_score(
                    config_df["gold"],
                    config_df["prediction"]
                )

                macro_f1 = f1_score(
                    config_df["gold"],
                    config_df["prediction"],
                    average="macro",
                    zero_division=0
                )

                records.append({
                    "feature": feature,
                    "task": task,
                    "bin": str(bin_value),
                    "config": config,
                    "n": len(config_df),
                    "accuracy": accuracy,
                    "macro_f1": macro_f1,
                    "feature_min": config_df[feature].min(),
                    "feature_max": config_df[feature].max(),
                    "feature_mean": config_df[feature].mean()
                })

bin_performance = pd.DataFrame(records)

print(bin_performance.shape)
display(bin_performance.head())
```

```python
def plot_feature_performance(
    feature,
    task,
    metric="macro_f1",
    figsize=(10, 5)
):

    d = bin_performance[
        (bin_performance["feature"] == feature) &
        (bin_performance["task"] == task)
    ].copy()

    if d.empty:
        print("No data:", feature, task)
        return

    # Preserve bin order
    bin_order = (
        d[["bin", "feature_min"]]
        .drop_duplicates()
        .sort_values("feature_min")["bin"]
        .tolist()
    )

    plt.figure(figsize=figsize)

    for config in d["config"].unique():

        x = d[d["config"] == config].copy()

        x["bin"] = pd.Categorical(
            x["bin"],
            categories=bin_order,
            ordered=True
        )

        x = x.sort_values("bin")

        plt.plot(
            x["bin"],
            x[metric],
            marker="o",
            linewidth=2,
            label=config
        )

    plt.xlabel(feature)
    plt.ylabel(metric.replace("_", " ").title())
    plt.title(
        f"{task}: {metric.replace('_', ' ').title()} "
        f"by {feature} quartile"
    )

    plt.xticks(rotation=30)
    plt.ylim(0, 1)
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()
```
